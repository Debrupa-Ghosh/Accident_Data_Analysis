```python
import numpy as np
```

```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
df=pd.read_csv("accident data.csv")
df
```

```python
df.head()
```

```python
df['Accident Date'] = pd.to_datetime(df['Accident Date'], errors='coerce') #If the column contains incorrect date formats, use errors='coerce' to convert invalid values to NaT (Not a Time).
```

```python
df
```

```python
df['Accident_Severity']=df['Accident_Severity'].str.lower().str.strip()
df['Light_Conditions']=df['Light_Conditions'].str.lower().str.strip()
df['Road_Type']=df['Road_Type'].str.lower().str.strip()
df['Vehicle_Type']=df['Vehicle_Type'].str.lower().str.strip()
df['Weather_Conditions']=df['Weather_Conditions'].str.lower().str.strip()
df['Road_Surface_Conditions']=df['Road_Surface_Conditions'].str.lower().str.strip()
df
```

```python
weather_counts = df['Weather_Conditions'].value_counts()

x = range(len(weather_counts))  # Convert categorical values to numbers
y = weather_counts.values

plt.plot(x, y, color='b', linestyle='-.', marker='o')

# Fix x-axis labels to show weather conditions instead of numbers
plt.xticks(ticks=x, labels=weather_counts.index, rotation=90)

plt.xlabel("Weather Condition")
plt.ylabel("Number of Accidents")
plt.title("Accidents by Weather Condition")

plt.show()
```

```python
df['Severity_Score'] = df['Accident_Severity'].replace({"Slight": 1, "Serious": 2, "Fatal": 3})

# Group by road type and calculate average severity
avg_severity = df.groupby('Road_Type')['Severity_Score'].mean()

plt.figure(figsize=(10, 5))
plt.plot(avg_severity.index, avg_severity.values, color='b',marker='o')
plt.title("Road Type vs Accident Severity")
plt.xlabel("Road Type")
plt.ylabel("Accident Severity")
plt.xticks()
for i,values in enumerate(avg_severity.values):
      plt.text(i, values+0.002, f"{values:.2f}", ha='center', va='bottom', fontsize=9)
plt.show()
```

```python

sns.countplot(x=df["Vehicle_Type"], hue=df["Accident_Severity"], palette=['black','green','red'])
plt.title("Accident Severity by Vehicle Type")
plt.xlabel("Vehicle Type")
plt.ylabel("Number of Accidents")
plt.xticks(rotation=90)
plt.legend(title="Severity")
plt.show()
```

```python
df_clean = df.dropna(subset=['Latitude', 'Longitude'])
```

```python
import folium
```

```python
from folium.plugins import MarkerCluster
```

```python
map_accidents = folium.Map(location=[df_clean["Latitude"].mean(), df_clean["Longitude"].mean()], zoom_start=6)
marker_cluster = MarkerCluster().add_to(map_accidents)

for _, row in df_clean.iterrows():
    folium.CircleMarker(
        location=[row["Latitude"], row["Longitude"]],
        radius=5,
        color="red" if row["Accident_Severity"].lower() == "severe" else "orange",
        fill=True,
        fill_color="red" if row["Accident_Severity"].lower() == "severe" else "orange"
    ).add_to(marker_cluster)

map_accidents.save("accident_hotspots.html")
```
